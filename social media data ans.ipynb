This data set is about social media, specifically about "social media sentiment & engagaement analytics"

negative,neutral)
rows--2260
columns--13


Column Description

1. post_id: A unique identification number assigned to each social media post.

2. timestamp: The date and time when the post was published.

3. platform: The social media platform where the post was shared.

4. user: The username of the person who created the post.

5. followers_count: The total number of followers of the user.

6. verified: Indicates whether the user's account is verified or not.

7. text: The actual text or content of the social media post.

8. hashtags: Hashtags used in the social media post.

9. country: The country associated with the user or post.

10. likes: The total number of likes received by the post.

11. retweets_shares: The total number of times the post was shared or retweeted.

12. comments: The total number of comments received by the post.

13. sentiment: Indicates the emotional tone of the post, such as Positive, Negative, or Neutral.

# Social Media Dataset - Data Cleaning Overview

This document summarizes the complete data cleaning and preprocessing workflow performed on the social media dataset to prepare it for analysis and visualization.

### 1. Initial Inspection & Structure (`df.shape`, `.info()`, `.dtypes`)
* Checked dataset dimensions using `df.shape` and inspected column structures, memory usage, and initial data types (`dtypes`) via `.info()`.

### 2. Header & Text Standardization (Capitalization)
* **Header Capitalization:** Capitalized and standardized column headers for better readability and consistency.
* **String & Categorical Standardization:** Stripped extra whitespace (`.str.strip()`) and unified inconsistent casing in categorical columns—specifically standardizing country names (e.g., `usa`, `Usa`, `USA` $\rightarrow$ `USA`) to prevent duplicate categories.

### 3. Data Type Conversion (`dtypes`)
* **Datetime Conversion:** Converted the `timestamp` column to `datetime64` format for accurate time-series analysis.
* **Numerical Formatting:** Converted engagement metrics (`followers_count`, `likes`, `retweets_shares`, and `comments`) into nullable integer (`Int64`) data types.

### 4. Handling Missing Values (`isnull`)
* Identified missing values across all columns using `.isnull().sum()`.
* **Numerical Columns:** Applied **median imputation** to fill missing values without distorting the data distribution.
* **Categorical & Text Columns:** Replaced null values with meaningful placeholder labels (e.g., `'Unknown'`, `'No Hashtag'`) to preserve all rows without data loss.

### 5. Duplicate Removal & Final Verification
* Identified and removed duplicate entries using `.duplicated()` and `.drop_duplicates()`.
* Re-verified final `dtypes`, `df.shape`, `.info()`, and `.head()` to confirm zero remaining null values and ensure the dataset is completely clean and ready for analysis.
